# **1. 지도학습과 강화학습**

### 1.  지도학습 (Supervised Learning)
지도학습은 정답(라벨)이 포함된 데이터로 모델을 학습시키는 방식입니다. 
<br>즉, 입력 X와 그에 대응하는 정답 Y를 함께 제공하여, 모델이 X→Y 매핑하는 함수를 학습하도록 합니다. 
<br>학습 과정에서는 모델의 예측값과 실제 정답 사이의 오차(loss)를 계산하고, 이를 최소화하도록 가중치를 업데이트합니다. 
<br>대표적인 문제 유형은 분류(Classification)와 회귀(Regression)이며, 이미지 분류, 스팸 메일 판별, 주가 예측 등이 이에 해당합니다. 
<br>핵심은 "정답이 있는 데이터로 오차를 줄이며 학습한다"는 점입니다.

### 2. 강화학습 (Reinforcement Learning) <- 어느정도 공부하느냐에 따라 성능차이가 심함. 
강화학습은 정답을 직접 주지 않고, 행동에 대한 보상(Reward)을 통해 학습하는 방식입니다. 
<br>에이전트(agent)가 환경(environment)과 상호작용하며 행동(action)을 선택하고, 그 결과로 받은 보상을 최대화하는 방향으로 정책(policy)을 학습합니다. 
<br>즉, "이 행동이 좋았는가?"를 보상으로 평가받으며 점점 더 좋은 전략을 찾아갑니다. 게임 AI, 자율주행, 로봇 제어 등이 대표적인 예입니다. 
<br>핵심은 "정답 대신 보상을 통해 최적 행동 전략을 학습한다"는 점입니다.

### 1. 보상(Reward)
- 에이전트가 특정 상태에서 어떤 행동을 했을 때 환경으로부터 받는 즉각적인 피드백 신호
- 행동의 좋고 나쁨을 수치로 표현한 값으로, 에이전트는 이 리워드를 기반으로 자신의 정책을 업데이트하면 장기적인 누적 보상을 최대화하도록 학습
- 매 타임스탭마다 주어질 수도 있고, 에피소드 종료 시 한 번만 주어질 수도 있으며, 양수는 바람직한 행동을, 음수는 바람직하지 않은 행동을 의미하는 경우가 일반적
    - 즉각적 피드백: 특정 상태에서 행동을 했을 때 즉시 주어지는 수치형 피드백
    - 희소성: 리워드는 매 순간 주어지지 않을 수 있음. 많은 환경에서 대부분의 시간 단계에서 보상이 0이고, 특정 목표를 달성했을 때만 보상이 주어질 수 있음 -> 희소할수록 학습이 어려워짐
    - 지연성: 어떤 행동의 결과에 대한 보상이 즉시 나타나지 않고, 여러 단계를 거친 귀 나중에 주어질 수 있음. 이로 인해 어떤 행동이 최종 결과에 기여했는지를 파악하기 어려워지며, 이를 신용할당 문제(Credit Assignment Probliem)라고 함
- 강화학습의 목표가 단순히 현재 리워드를 최대화하는 것이 아니라, 할인율을 고려한 미래 보상까지 포함한 누적 리워드를 최대화하는 것이 중요


### 2. 에이전트(Agent)
- 환경과 상호작용하며 상태를 관찰하고, 그에 따라 행동을 선택하고, 그 결과로 리워드를 받아 학습하는 주체를 의미
- 에이전트의 목표는 단기적인 보상이 아니라 미래 보상을 포함한 누적 보상을 최대화하는 정책을 학습하는 것
- 경험을 통해 가치 함수나 정책을 점진적으로 업데이트하며, 탐험과 활용을 균형 있게 수행

1. 가치 기반 에이전트
    - 가치 함수를 학습하여 행동을 선택하는 방식
    - 각 상태 또는 상태-행동 쌍의 가치를 계산하고, 가치가 가장 높은 행동을 선택
2. 정책 기반 에이전트
    - 행동 선택 규칙인 정책을 직접 학습하는 방식
    - 확률적으로 행동을 출력하는 정책을 최적화함
3. 액터-크리틱 에이전트
    - 가치 기반과 정책 기반을 결합한 방식
    - Actor는 정책을 학습하여 행동을 선택
    - Critic은 가치 함수를 학습해 Actor의 행동을 평가

### 3. 탐험(Exploration)
- 탐험이란, 에이전트가 현재까지 학습한 최적 행동만 반복하는 것이 아니라, 더 나은 보상을 얻을 가능성이 있는 새로운 행동을 시도해보는 과정을 의미
- 장기적인 누적 보상을 최대화하기 위해 필수적인 전략으로, 이미 잘 알려진 행동을 선택하는 활용과 균형을 이루어야 함

### 4. 활용(Exploitation)
- 활용이란, 에이전트가 지금까지의 학습 결과를 바탕으로 현재 가장 높은 보상을 줄 것으로 예상되는 행동을 선택하는 과정을 의미
- 이미 경험을 통해 가치가 높다고 판단된 행동을 반복하여 안정적으로 보상을 얻는 전략
- 단기적으로는 높은 성과를 보장하지만, 새로운 가능성을 탐색하지 않기 때문에 더 나은 전략을 발견하지 못하고 지역 최적해에 머무를 위험이 있음

# **2. 강화학습을 게임에 대입**

### 1. 주요 용어

1. Agent: 선택하는 주체. 캐릭터
2. Environment: 행동에 반응하는 세계. 미로와 벽
3. State: 의사결정에 필요한 현재 정보. 캐릭터의 위치, 열쇠 보유 여부
4. Observation: 실제로 볼 수 있는 정보. 화면의 일부
5. Action: 선택 가능한 조작. 위/아래/좌/우
6. Reward: 한 단계에서 받은 점수. 도착(+10), 충돌(-1)
7. Policy: 상황별 행동 규칙. 갈림길에서 오른쪽 선택
8. Return: 앞으로 받을 보상들의 합. 여러 이동의 총 평가 점수
9. Episode: 시작부터 종료까지 한 판. 출판 -> 도착/실패

### 2. 보상
- 즉시 보상: 동전을 주우면 +1
- 지연 보상: 바둑은 여러 수를 둔 뒤 승리했을 때 +1
- 희소 보상: 미로 출구에 도착하기 전까지 0점
- 보상 설계의 함정: 앞으로 잘 이동할 때마다 +1을 주면 출구 대신 계속 왔다갔다 이동만 할 수 있음
> 순차적 의사결정: 지금 장애물을 피하기 위해 돌아가는 행동은 당장은 느리지만, 나중에 목적지에 안전하게 도착할 수 있음
<br>ε-greedy: 10%는 무작위로 탐험. 나머지 90%는 현재 가장 좋다고 추정한 행동 선택. 최고는 지금까지의 추정치이지 진짜 최고가 아님

- 에이전트는 슬롯머신을 직접 선택하고, 당첨 여부를 확인하면서 각 머신의 당첨 확률을 추정해야 함

2. AI 에이전트 구현
- Random 전략: 5개의 슬롯머신 중 하나를 무작위로 선택
- Greedy 전략: 지금까지 획득한 보상을 기반으로 가장 높은 평균 보상을 가진 슬롯머신을 선택
- ε-greedy 전략: 일정 확률로 무작위 슬롯머신을 선택하고, 나머지 확률로 현재 가장 높은 평균 보상을 가진 슬롯머신을 선택(ε값은 0.1과 0.01을 각각 적용)

3. 학습 기능 구현
AI 에이전트는 슬롯머신을 선택할 때마다 아래 정보를 업데이트 해야 함
- 각 슬롯머신을 선택한 횟수
- 각 슬롯 머신의 평균 보상 추정값
- 매 행동에서 획득한 보상
- 지금까지 획득한 누적 보상
> 평균 보상은 기존 데이터를 모두 다시 계산하지 않고, 이전 평균과 새로 획득한 보상을 이용하는 방식으로 구현

# **4. 전략별 실험**
각 전략을 1000번씩 실행하고 아래 결과를 출력
- 최종 누적 보상
- 슬롯머신별 선택 횟수
- 슬롯머신별 추정 당첨 확률
- 가장 많이 선택한 슬롯머신 번호

````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
TRUE_PROBS = np.array([0.2, 0.5, 0.1, 0.8, 0.3])  # 0~4번 슬롯머신의 성공 확률
N_STEPS = 1000  # 총 선택 횟수

# 전략(랜덤, 그리디, 엡실론-그리디)별로 얼마나 보상을 받는지 시뮬레이션하는 함수
def bandit_experiment(strategy="epsilon", epsilon=0.1, seed=2026):
    rng = np.random.default_rng(seed) 
    counts = np.zeros(5, dtype=int)  # 선택 횟수용 5칸을 정수 0으로 채움
    estimates = np.zeros(5, dtype=float)  # 추정 확률용 5칸을 실수 0.0으로 채움
    rewards = []  # 보상을 담을 빈 리스트
    choices = []  # 선택 번호를 담을 빈 리스트

    for t in range(N_STEPS):
        if t < 5:
            action = t  # 처음엔 0~4번을 한 번씩 선택
        elif strategy == "random":
            action = int(rng.integers(5))  # 0~4 중 무작위 선택
        elif strategy == "greedy":
            action = int(np.argmax(estimates))  # 추정값이 가장 큰 번호 선택
        elif strategy == "epsilon":
            action = int(rng.integers(5)) if rng.random() < epsilon else int(np.argmax(estimates))  # epsilon 확률로 무작위, 그 외엔 최대 추정값 선택
        else:
            raise ValueError("unknown strategy")  # 모르는 전략이면 오류

        reward = float(rng.random() < TRUE_PROBS[action])  # 난수가 성공 확률보다 작으면 1.0, 아니면 0.0
        counts[action] += 1  # 해당 슬롯머신의 선택 횟수 +1
        estimates[action] += (reward - estimates[action]) / counts[action]  # 새 보상을 반영해 평균 갱신
        rewards.append(reward)  # 이번 보상을 맨 뒤에 추가
        choices.append(action)  # 이번 선택 번호를 맨 뒤에 추가
    return np.cumsum(rewards), counts, estimates  # 누적 보상, 선택 횟수, 추정 확률
````

````python
results = {
    "Random": bandit_experiment("random"),
    "Greedy": bandit_experiment("greedy"),  # 추정값이 가장 큰 번호 선택
    "ε=0.1": bandit_experiment("epsilon", 0.1),  # 10% 확률로 무작위 탐색
    "ε=0.01": bandit_experiment("epsilon", 0.01),  # 1% 확률로 무작위 탐색
    "ε=0": bandit_experiment("epsilon", 0) # 강사님이 얘기해서 내가 해봄

}
````

````python
# SEED를 바꿔가면서 보고 평균을 내봐야 뭐가 좋은지 안다. 하지만 현재는 0.01이 제일 좋음 흠 나중에 SEED는 지우고 하는게 좋지 않을까 싶다
for name, (cumulative, counts, estimates) in results.items():
    print(f"{name:8s} | 누적 보상={cumulative[-1]:.0f} | 선택 횟수={counts} | 추정={estimates.round(2)}")

plt.figure(figsize=(9, 4))


for name, (cumulative, _, _) in results.items():
    plt.plot(cumulative, label=name)
plt.xlabel("Number of actions")
plt.ylabel("Cumulative reward")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
````

# **4. 마르코프 성질**
"현재 상태와 현재 행동을 알면 다음 상태의 확률을 설명하는 데 과거 기록이 추가로 필요하지 않다."
- 어떤 확률 과정에서 미래 상태가 과거 전체의 이력에 의존하지 않고 오직 현재 상태에만 의존함
- 현재 상태가 과저 정보를 모두 요약하고 있다는 의미

### MP(Markov Process)
- 행동을 선택하는 주체가 없다는 것

```
오늘 날씨가 맑으면 내일은 아래와 같이 변한다고 가정
- 맑음 > 맑음(70%)
- 맑음 > 비(30%)
여기에는 누군가가 선택하는 행동이 없음
```
> 현재 상태에 따라 다음 상태가 확률적으로 결정

```
MP = (S, P)
- S: 상태(맑음, 비)
- P: 상태 전이확률
```

### MRP(Markov Reward Process)
- MP에 보상이라는 개념을 추가한 것

```
맑음 -> 맑음 -> 비 -> 맑음
+10     +10     -10    +10
```

```
MRP = (S, P, R, γ)
- S: 상태
- P: 전이확률
- R: 보상
- γ: 할인율. 미래에 받을 보상을 현재 기준으로 얼마나 중요하게 평가할 것인지
```

### MDP(Markov Decision Process)
- MRP와의 큰 차이는 에이전트가 직접 행동을 선택할 수 있다는 것
- MRP에 행동 선택이 추가되어 정책을 비교할 수 있음
- 기존 마르코드 보상 과정에서 행동이 추가된 확률 모델 <- 내가 책보고 넣음
<BR><BR>

```
기대 보상이 높아서 위로 이동이 더 좋지 않을까
```
| 행동 | 성공확률 | 실패확률 | 성공보상 | 실패보상 |
|:---:|:---:|:---:|:---:|:---:|
| 위로 이동 | 80% | 20% | +10 | -10 |
| 아래로 이동 | 60% | 40% | +5 | -5 |
```
0.8(10) + 0.2(-10) = 6
0.6(5) + 0.4(-5) = 1
```

> 위와 같은 환경에서 어떤 행동을 선택해야 미래에 많은 보상을 얻을 수 있는지 학습하는 것이 강화학습의 핵심

### 전이 확률

```
마르코프 체인은 시간에 따른 상태 변화를 나타내며 이 상태 변화를 전이라고 함
마르코프 프로세스에서 상태 간 이동이 전이는 확률로 표현하게 되는데. 이를 상태 전이 확률이라 함
즉, 시간 t에서의 상태를 s라고 하며, 시간 t+1에서의 상태를 s'라고 할 때 다음과 같은 수식으로 표현할 수 있음

수식: P(S_(t+1) = s' | S_t = S) = P_(ss')

```


```
로봇이 "오른쪽 이동"을 명령했을 때 
- 80%: 오른쪽으로 이동
- 10%: 미끄러져 위쪽으로 이동
- 10%: 제자리 유지
```
> 합계가 100%여야 함. 행동을 선택하더라도 다음 상태가 항상 확정되는 것은 아님

````python
# 전이 확률: "오른쪽 이동" 명령을 내렸을 때 실제로 일어나는 결과와 그 확률
transition = {"right": 0.8, "up": 0.1, "stay": 0.1}

# 확률의 합이 1(100%)인지 확인. 아니면 오류 발생
# (부동소수점 오차 때문에 == 대신 np.isclose 사용)
assert np.isclose(sum(transition.values()), 1.0)

rng = np.random.default_rng(7)  # seed 7로 난수 고정 → 실행할 때마다 같은 결과

# 위 확률대로 10000번 뽑기
# list(transition): ["right", "up", "stay"] (뽑을 후보)
# p=list(transition.values()): [0.8, 0.1, 0.1] (각 후보의 확률)
samples = rng.choice(list(transition), size=10000, p=list(transition.values()))

# 각 결과가 실제로 몇 %나 나왔는지 출력
for next_state in transition:
    # samples == next_state: 해당 결과면 True(1), 아니면 False(0)인 배열
    # np.mean: 그 평균 = 해당 결과가 나온 비율
    # 3은 소수점
    print(next_state, "실험 비율:", round(np.mean(samples == next_state), 3))
````

### 할인율 γ와 Return

```
게임에서 지금 +10점, 다음 턴 +10점, 그다음 턴 +10점을 받음
- γ=1: 10 + 10 + 10 = 30점(모두 같은 비중)
- γ=0.9: 10 + 9 + 8.1 = 27.1점
- γ=0.5: 10 + 5 + 2.5 = 17.5점
```

> γ는 "미래의 보상을 현재 평가에서 얼마나 중요하게 볼지"를 정함. γ기 1에 가까우면 미래를 더 오래 고려한다는 뜻이지, 미래가 현재보다 더 중요하다는 뜻은 아님

````python
rewards = [10, 10, 10]  # 3번의 턴에서 받은 보상 (R₁, R₂, R₃)

# 할인율 γ를 바꿔가며 리턴이 어떻게 달라지는지 비교
for gamma in [0.0, 0.5, 0.9, 1.0]:
    # G = R₁ + γR₂ + γ²R₃ + ...
    # enumerate: (t, reward) = (0, 10), (1, 10), (2, 10)
    # gamma ** t: t번째 보상에 곱할 할인 비율 (먼 미래일수록 더 많이 깎임)
    discounted = sum((gamma ** t) * reward for t, reward in enumerate(rewards))

    # :.1f → 소수점 1자리, :.2f → 소수점 2자리까지 출력
    print(f"gamma={gamma:.1f} -> return={discounted:.2f}")
````

- 에이전트는 working(공부), gym(운동), home(귀가) 세 가지 상태를 가지며, 선택한 행동에 따라 상태가 변경되고 보상을 받음

| working | gym | gym | +5 |

### 구현 요구사항
- 상태 전이와 보상을 저장하는 딕셔너리를 작성
- 행동 목록과 할인율을 전달받는 함수를 구현
- 시작 상태는 working으로 설정
- 행동을 순서대로 실행하며 상태를 변경하고 보상을 저장
- 불가능한 행동을 선택하거나 종료 상태에서 추가 행동을 실행하면 오류가 발생하도록 구현
- 아래 공식을 사용하여 할인 리턴을 계산

유니코드: G = R₁ + γR₂ + γ²R₃ + ⋯

$G = R_1 + \gamma R_2 + \gamma^2 R_3 + \cdots$

- 할인율을 기본적으로 0.9를 사용
- 두 가지 경로를 실행하고 결과를 비교
    - working -> working -> home
    - gym -> working -> home

````python
# 내꺼
start = "working"

transP_dict = {
    ("working", "working"): ("working", 3),
    ("working", "gym"):     ("gym", 5),
    ("working", "home"):    ("home", 12),
    ("gym", "working"):     ("working", 4),
    ("gym", "home"):        ("home", 12),
}




````

````python
TRANSITIONS = {
    ("working", "working"): ("working", 3),
    ("working", "gym"):     ("gym", 5),
    ("working", "home"):    ("home", 12),
    ("gym", "working"):     ("working", 4),
    ("gym", "home"):        ("home", 12),
}

def evaluate_actions(actions, gamma=0.9):
    state, rewards = "working", []
    for action in actions:
        if state == "home":
            raise ValueError("종료 상태 이후 행동이 있습니다")

        if(state, action) not in TRANSITIONS:
            raise ValueError(f"불가능한 행동: {state} -> {action}")
        state, reward = TRANSITIONS[(state, action)]
        rewards.append(reward)

    if state != "home":
        print("주의: 아직 종료되지 않은 경로입니다.")
    return rewards, sum(gamma ** t * r for t, r in enumerate(rewards))
````

````python
for name, actions in {
    "A": ["working", "working", "home"],
    "B": ["gym", "working", "home"]
}.items():
    rewards, G = evaluate_actions(actions)
    print(f"정책 경로 {name}: 보상={rewards}, 할인 리턴={G:.2f}")
````

# **6. 벨만 기대 방정식**
현재 상태의 가치를 계산할 때, 지금 받을 보상과 다음 상태에서 앞으로 받을 보상의 예상 가치를 합치는 공식
1. 학생이 게임하고 있다고 가정
- 보상: 특정 행동을 했을 때 즉시 얻는 점수
- 리턴: 지금부터 앞으로 얻을 보상들을 할인해서 합한 값
- 가치: 현재 상태에서 앞으로 얻을 리턴의 기댓값
```
책
어떤 상황에서 미래의 보상을 포함한 가치를 나타냄. 다음 상태로 이동하려면 어떤 정책에 따라 행동해야 하는데 이때 정책을 고려한 다음 상태로의 이동이 벨만 기대 방정식
이 방정식으로  상태- 가치 함수와 행동-가치 함수를 기댓값 E로 표현할 수 있음
즉, 재귀적인 형태로서 미래의 가치들이 현재의 가치에 영향을 주고 있는 형태라고 이해하면 됨
```
```
        +2      +10
    시작 -> 중간 -> 종료

    단순 보상 합계: 2 + 10 = 12
    할인율(0.9) 적용: 2 + 0.9 * 10 = 11
```

- 벨만 방정식이 필요한 이유
    - 위와 같은 예제는 보상이 두 번밖에 없어서 직접 더할 수 있음
    - 만약 게임이 100번, 1000번 진행된다면 매번 미래의 모든 보상을 직접 나열하는 것은 어려움
        - 벨만 방정식으로 해결
        - 현재 가치 = 지금 받을 보상 + 할인된 다음 상태의 가치
        - $V(s) = R + γV(s')$, 유니코드: V(s) = R + γV(s′)

```
    V(A) = 10
    V(S) = 2 + 0.9 * V(A)
    = 2 + 0.9 * 10 = 11
```
> 미래 보상을 전부 다시 계산하지 않고, 다음 상태의 가치를 이용해 현재 가치를 계산하는 것

- 다음 상태가 확률적으로 달라질 경우
```
    게임에서 오른쪽 버튼을 눌렀는데, 항상 같은 칸으로 이동하지 않는다고 가정
    - 50% 확률로 좋은 칸에 이동
    - 50% 확률로 나쁜 칸에 이동
    - 좋은 칸의 미래 가치는 10점, 나쁜 칸의 미래 가치는 -2점

    기댓값을 계산
    10 * 0.5 + (-2) * 0.5 = 5 - 1 = 4(예상 가치)

    현재 보상 0이고, 할인율이 0.9라면
        0 + 0.9 * 4 = 3.6
    - 여기서 3.6은 실제로 받는 점수가 아님
    - 좋은 칸에 도착하면 할인된 미래 가치는 9점이고, 나쁜 칸에 도착하면 -1.8점 -> 두 결과를 확률에 따라 평균내면 3.6이 됨
```


$\boxed{V^\pi(s)=\sum_a\pi(a|s)\sum_{s'}P(s'|s,a)[R(s,a,s')+\gamma V^\pi(s')]}$

$R+\gamma V(s')$


$\sum_{s'}P(s'|s,a)[R+\gamma V(s')]$

$\sum_a\pi(a|s)(\text{행동별 기대 가치})$

````python
good_value, bad_value = 10.0, -2.0
gamma = 0.9
expected_next = 0.5 * good_value + 0.5 * bad_value
print("다음 상태 가치의 평균:", expected_next)
print("지금 상태에서 기대하는 가치:", 0 + gamma * expected_next)
````

````python
STATES = ["Home", "A", "B", "C", "Bar"]
NEXT = {
    "A": [("Home", 0.7), ("B", 0.3)],
    "B": [("A", 0.5), ("C", 0.5)],
    "C": [("B", 0.5), ("Bar", 0.5)]
}
TERMINAL = {"Home", "Bar"}
````

````python
def arrival_reward(next_state):
    return 1.0 if next_state == "Home" else -1.0 if next_state == "Bar" else 0.0
````

````python
# 상태 가치 함수를 저장
V = {state: 0.0 for state in STATES}

for iteration in range(1, 101):
    new_V = V.copy()

    # state: 현재 상태
    # nxt: 다음 상태
    # prob: 전이 확률
    # arrival_reward(nxt): 이동 시 보상
    # V[nxt]: 다음 상태의 가치
    for state in ["A", "B", "C"]:
        new_V[state] = sum(prob * (arrival_reward(nxt) + V[nxt]) for nxt, prob in NEXT[state]) # 벨만 기대 방정식

    # 이전 반복과 현재 반복의 상태 가치가 얼마나 달라졌는지 비교
    delta = max(abs(new_V[s] - V[s]) for s in STATES)

    # 이전 반복에서 계산한 새로운 상태 가치를 다음 반복에서 사용할 수 있도록 갱신
    V = new_V


    if iteration in [1, 2, 5, 10, 20, 50, 100]:
        print(f'{iteration:3d}회: A={V['A']:+.4f}, B={V['B']:+.4f}, C={V['C']:+.4f}')
    if delta < 1e-12:
        break

# assert: 특정 조건이 참인지 확인
# allclose: 두 배열의 값이 허용 오차 범위 내에서 거의 같은지 확인
# atol: 절대 허용 오차를 설정
# [0.75, 1/6, -5/12] <- 값이 이게 아니면 오류남
assert np.allclose([V['A'], V['B'], V['C']], [0.75, 1/6, -5/12], atol=1e-6) # atol 오차 허용 범위
````

# **8. 몬테카를로 방법**
몬테카를로(Monte Carlo) 방법은 무작위 표본을 반복적으로 생성하고, 그 결과의 평균을 이용하여 어떤 값의 기댓값을 추정하는 방법

| 실행 | 최종 도착 | 보상 |
|:---:|:---:|:---:|
| 1 | Home | +1 |
| 2 | Home | +1 |
| 3 | Bar | -1 |
| 4 | Home | +1 |
| 5 | Home | +1 |
```
평균 보상: 0.6
```
> 5번만 실행했기 때문에 정확하지 않을 수 있지만, 실험 횟수를 10000번 정도로 늘리면 평균값은 이 문제의 이론적 가치인 0.5에 가깡워지는 경향이 있음

$V_\pi(s)\approx\frac{1}{N(s)}\sum_{i=1}^{N(s)}G_i(s)$

$G_t=R_{t+1}+\gamma R_{t+2}+\gamma^2R_{t+3}+\cdots$

# 벨만 방정식과 몬테카를로의 차이
1. 벨만 방정식
- 계산 원리: 전이 확률로 기대값 계산
- 전이 확률: 필요
- 에피소드: 이무진 실행할 필요 없음
- 결과: 수렴한 상태 가치
- 특징: 모델 기반


2. 몬테카를로
- 계산 원리: 실제 경험의 리턴 평균
- 전이 확률: 필요 없음
- 에피소드: 종료까지 실행
- 결과: 경험 기반 추정 가치
- 특징: 모델 프리


````python
def random_walk_episode(start, rng):
    state = start
    while state not in TERMINAL:
        options, probs = zip(*NEXT[state])
        state = str(rng.choice(options, p=probs))
    return arrival_reward(state)

# 위에 이거 반영함
# def arrival_reward(next_state):
#     return 1.0 if next_state == "Home" else -1.0 if next_state == "Bar" else 0.0
````

````python
SIZE = 5
HOME = (0, 0)
BAR = (0, 4)
GAMMA = 0.95
ACTIONS = {
    "up": (-1, 0),
    "down": (1, 0),
    "left": (0, -1),
    "right": (0, 1)
}
ACTION_NAMES = ['up', 'down', 'left', 'right']
ARROWS = {"up": "↑", "down": "↓", "left": "←", "right": "→", None: "·"}
````

````python
def grid_step(state, action):
    if state in (HOME, BAR):
        return state, 0.0, True

    dr, dc = ACTIONS[action]

    row, col = state
    next_state = (min(max(row+dr, 0), SIZE-1), min(max(col+dc, 0), SIZE-1))

    if next_state == HOME:
        reward = 1.0
    elif next_state == BAR:
        reward = -1.0
    else:
        reward = 0.0
    done = next_state in (HOME, BAR)
    return next_state, reward, done
````

````python
def show_values(values, digits=3):
    print(np.array2string(np.round(values, digits), formatter={"float_kind": lambda x: f"{x:7.{digits}f}"}))
````

````python
def show_policy(policy):
    for row in range(SIZE):
        print(" ".join(ARROWS[policy[row, col]]for col in range(SIZE)))
````

````python
samples = [
    ((1, 0), "up"),
    ((0, 1), "left"),
    ((0, 3), "right"),
    ((2, 0), "left"),
    ((0, 0), "down")
]

for state, action in samples:
    nxt, reward, done = grid_step(state, action)
    print(f"{state} + {action:5s} -> {nxt}, 보상={reward:+.0f}, 종료={done}")
````

````python
def evaluate_random_policy(gamma=GAMMA, tol=1e-10, max_iter=10000):
    values = np.zeros((SIZE, SIZE))
    for iteration in range(max_iter):
        updated = values.copy()
        for row in range(SIZE):
            for col in range(SIZE):
                state = (row, col)
                if state in (HOME, BAR):
                    continue
                action_values = []
                for action in ACTION_NAMES:
                    nxt, reward, done = grid_step(state, action)
                    future = 0.0 if done else values[nxt]
                    action_values.append(reward + gamma * future)
                updated[state] = np.mean(action_values)
        delta = np.max(np.abs(updated - values))
        values = updated
        if delta < tol:
            return values, iteration + 1
    raise RuntimeError("랜덤 정책 평가가 수렴하지 않음")
````

````python
random_values, random_iters = evaluate_random_policy()
print("gamma=0.95 반복 횟수: ", random_iters)

show_values(random_values)
assert random_values[HOME] == 0 and random_values[BAR] == 0

low_values, low_iters = evaluate_random_policy(gamma=0.5)
print("gamma=0.5 반복 횟수: ", low_iters)
show_values(low_values)
````

# **10. 정책 개선**
- 랜덤 정책의 가치는 "마구 걸으면 이 칸이 얼마나 좋은가?"에 대한 방법
- 해당 숫자를 이용하면 각 칸에서 더 나은 화살표를 고를 수 있음
- Q(s, a) = r + γ * (종료면 0, 아니면 V(s'))
    - s: state
    - a: action
- 개선된 정책은 각 칸에서 이 Q가 가장 큰 행동을 고름. 점수가 같으면 up, down, left, right 중 먼저 나온 행동을 고름

````python
# 한 상태에서 4방향 각각의 행동 점수(R + γV(s'))를 계산해 리스트로 반환
def action_scores(state, values, gamma=GAMMA):
    scores = []  # up, down, left, right 순서로 점수를 담음
    for action in ACTION_NAMES:
        nxt, reward, done = grid_step(state, action)
        future = 0.0 if done else values[nxt]  # 종료면 미래 가치 없음
        scores.append(reward + gamma * future)  # R + γV(s')
    return scores
````

````python
# 가치 함수 values를 보고 각 칸에서 점수가 가장 높은 행동을 고르는 그리디 정책 생성
def greedy_policy_from(values, gamma=GAMMA):
    policy = np.empty((SIZE, SIZE), dtype=object)  # 칸마다 행동 이름(문자열)을 담을 배열
    
    for row in range(SIZE):
        for col in range(SIZE):
            state = (row, col)
            if state in (HOME, BAR):
                policy[state] = None  # 종료 상태는 행동이 없음 (show_policy에서 · 로 표시)
            else:
                # 4방향 점수 중 최대인 행동의 인덱스를 찾아 행동 이름으로 변환
                policy[state] = ACTION_NAMES[int(np.argmax(action_scores(state, values, gamma)))]
    return policy
````

# **11. 정책 반복**
정책 반복은 두 단계를 정책이 변하지 않을 때까지 반복
1. 정책 평가: 지금 화살표를 고정하고, 그 화살표만 따르는  V가 수렴할 때까지 갱신
2. 정책 개선: 그 V로 각 칸의 화살표를 다시 그림

> 평가 -> 개선 -> 비교 -> 다시 평가(반복)

````python
# 주어진 (결정적) 정책을 따랐을 때의 상태 가치를 반복 계산 (정책 평가)
def evaluate_policy(policy, gamma=GAMMA, tol=1e-10, max_iter=10000):
    values = np.zeros((SIZE, SIZE))  # 모든 상태 가치를 0으로 초기화
    for iteration in range(max_iter):
        updated = values.copy()  # 이전 값으로 계산하기 위해 복사본에 갱신
        for row in range(SIZE):
            for col in range(SIZE):
                state = (row, col)
                if state in (HOME, BAR):
                    continue  # 종료 상태 가치는 0 유지
                nxt, reward, done = grid_step(state, policy[state])  # 정책이 정한 행동 하나만 실행
                future = 0.0 if done else values[nxt]  # 종료면 미래 가치 없음
                updated[state] = reward + gamma * future  # R + γV(s'), 행동이 하나라 평균이 필요 없음
        delta = np.max(np.abs(updated - values))  # 이전 반복과의 최대 가치 차이
        values = updated  # 갱신된 값을 다음 반복에서 사용
        if delta < tol:  # 수렴 판정
            return values, iteration + 1
    raise RuntimeError("정책 평가가 수렴되지 않았음")
````

````python
# 두 정책이 모든 칸에서 같은 행동을 고르는지 비교 (정책 반복의 종료 조건에 사용)
def policies_equal(left, right):
    return all(left[r, c] == right[r, c] for r in range(SIZE) for c in range(SIZE))  # 모든 칸이 같아야 True
````

````python
# 정책 평가와 정책 개선을 번갈아 반복해 정책이 더 이상 안 바뀔 때까지 수렴시킴 (정책 반복)
def policy_iteration(gamma=GAMMA, max_outer=20):
    policy = np.full((SIZE, SIZE), 'right', dtype=object)  # 초기 정책: 모든 칸에서 right
    policy[HOME] = None  # 종료 상태는 행동 없음
    policy[BAR] = None
    for outer in range(1, max_outer + 1):
        values, eval_iters = evaluate_policy(policy, gamma=gamma)  # 1. 현재 정책의 가치 평가
        improved = greedy_policy_from(values, gamma=gamma)  # 2. 그 가치로 그리디 정책 개선
        print(f"바깥 {outer}회 | 평가 반복={eval_iters}")
        show_policy(improved)  # 개선된 정책을 화살표로 출력
        if policies_equal(policy, improved):  # 정책이 안 바뀌면 수렴
            return values, improved, outer  # 최적 가치, 최적 정책, 바깥 반복 횟수
        policy = improved  # 바뀌었으면 새 정책으로 다시 평가
    raise RuntimeError("정책 반복이 끝나지 않음")
````

````python
pi_values, pi_policy, pi_rounds = policy_iteration()
# print('정책 반복 최종 가치')
# # show_policy(pi_values)

# show_policy(pi_policy)
````
